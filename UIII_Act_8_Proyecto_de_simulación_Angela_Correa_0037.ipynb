### Ejercicio 14 Sastre : aguja 🧵
➡ El programa "Sastre y Aguja" es un asistente interactivo que simula una inteligencia artificial básica mediante lógica de programación pura.


```python
print("--- 🧵 BIENVENIDO AL TALLER 'SASTRE Y AGUJA' 🧵 ---")
continuar = "si"

while continuar == "si":
    print("\n¿En qué te puedo asesorar hoy?")
    print("1. Elegir tela según el clima")
    print("2. Selección de aguja")
    print("3. Salir")

    opcion = input("Elige una opción (1-3): ")

    if opcion == "1":
        clima = input("¿El diseño es para clima 'frio' o 'calor'?: ").lower()
        if clima == "frio":
            print("🤖 IA Sastre: Te recomiendo usar Lana o Pana para mantener el calor.")
        elif clima == "calor":
            print("🤖 IA Sastre: El Lino o Algodón fresco serán tus mejores aliados.")
        else:
            print("🤖 IA Sastre: No reconozco ese clima, pero la Mezclilla es universal.")
    elif opcion == "2":
        tipo_tela = input("¿La tela es 'delgada', 'media' o 'gruesa'?: ").lower()
        if tipo_tela == "delgada":
            print("🤖 IA Sastre: Usa aguja punta de bola 70/10.")
        elif tipo_tela == "media":
            print("🤖 IA Sastre: La aguja universal 90/14 es la ideal.")
        elif tipo_tela == "gruesa":
            print("🤖 IA Sastre: Necesitas fuerza, usa una 100/16.")
        else:
            print("🤖 IA Sastre: Ante la duda, usa una aguja universal nueva.")
    elif opcion == "3":
        continuar = "no"
        print("¡Gracias por usar 'Sastre y Aguja'! ¡Feliz costura!")
    else:
        print("❌ Opción no válida, intenta de nuevo.")
print("Programa realizado por Angela Michelle Correa Lopez NL = 14")
```
